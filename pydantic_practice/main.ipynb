```python
from prompt_toolkit.filters import app
from pydantic import BaseModel
from fastapi import FastAPI
class Food(BaseModel):
    name:str
    price:float
    available:bool=True
app=FastAPI()
@app.post("/foods")
def create_foods(food:Food):
    return {
        "name":food.name,
        "price":food.price,
        "available":food.available
    }

```

```python
# Custom validation with field validator

from pydantic import BaseModel,field_validator

class Food(BaseModel):
    name:str

    @field_validator('name')
    @classmethod
    def validate_name(cls,value:str)->str:
        value=value.strip()

        if not value:
            raise ValueError("Name cannot be empty")
        return value
```

```python
# Enum -Restrict the values to allowed choice

from enum import Enum
from pydantic import BaseModel

class FoodCategory(str,Enum):
    VEG:"veg"
    VEGAN:"vegan"

class Food(BaseModel):
    name:str
    category:FoodCategory
```

```python
from pydantic import BaseModel,Field

class FoodCreate(BaseModel):
    name:str
    price:float=Field(gt=0)

class FoodUpdate(BaseModel):
    name: str | None = None
    price: float | None = Field(default=None, gt=0)

class FoodResponse(BaseModel):
    id: int
    name: str
    price: float
```

```python
# Model Validator

from pydantic import BaseModel,model_validator

class Booking(BaseModel):
    start_hour:int
    end_hour:int

    @model_validator(mode="after")
    def validate_hour(self):
        if self.end_hour<=self.start_hour:
            raise ValueError("End hour cannot be greater than Start hour")
        return self
```
